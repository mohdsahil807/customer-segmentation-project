```python
import pandas as pd

df = pd.read_csv("../data/intermediate/merged_orders.csv")
df = df.copy()
print("shape:", df.shape)
```

```python
df.head()
```

```python
missing = df.isnull().sum()
missing_percent = (missing / len(df)) * 100

missing_summary = pd.DataFrame({
    "Missing Count": missing,
    "Missing %": missing_percent.round(2)
})

missing_summary = missing_summary[missing_summary["Missing Count"] > 0].sort_values(by="Missing %", ascending=False)
print(missing_summary)
```

```python
print(df.columns.tolist())
```

```python
# Sanity Check - Skip if columns have already been dropped.
cols_to_drop = ["review_comment_title", "review_comment_message"]
existing_cols = [col for col in cols_to_drop if col in df.columns]
if existing_cols:
    df = df.drop(columns=existing_cols)

# Fill the category missing with "unknown"
df["product_category_name"] = df["product_category_name"].fillna("unknown")
df["product_category_name_english"] = df["product_category_name_english"].fillna("unknown")

# By dropping 3 missing rows of payment
df = df.dropna(subset=["payment_value"])

print("New shape after cleaning:", df.shape)
```

```python
# Date columns convert by Datatime 

date_columns = [
    "order_purchase_timestamp",
    "order_approved_at",
    "order_delivered_carrier_date",
    "order_delivered_customer_date",
    "order_estimated_delivery_date",
    "shipping_limit_date",
    "review_creation_date",
    "review_answer_timestamp"
]

for col in date_columns:
    df[col] = pd.to_datetime(df[col],errors="coerce")

print("Date columns converted!")
df[date_columns].dtypes
```

```python
# Check duplicate rows

duplicate_count = df.duplicated().sum()
print("Total duplicate rows:", duplicate_count)
```

```python
df.to_csv("../data/intermediate/cleaned_orders.csv", index=False)
print("Cleaned data saved successfully!")
print("Final shape:",df.shape)
```
